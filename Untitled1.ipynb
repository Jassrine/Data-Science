```python
import numpy as np
import pandas as pd
np.random.seed(32)
data={
    'x':np.arange(1,6),
    'y':np.random.randint(10,100,size=5)
}
df=pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(3,3))
plt.plot(df)

plt.xlabel('day')
plt.ylabel('sale')
plt.title('graphs')
plt.show()
```

```python

import matplotlib.pyplot as plt
plt.figure(figsize=(3,3))
plt.plot(df)
plt.plot('df', marker='d',color='purple', linestyle='--')
plt.grid(True)
plt.xlabel('day')
plt.ylabel('sale')
plt.title('graphs')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0, 11)
y = x
plt.figure(figsize=(3,3))
plt.plot(x, y, marker='o', color='blue')   # using marker & color as you learned earlier
plt.grid(True)
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Basic Line Graph')
plt.show()
```

```python
import matplotlib.pyplot as plt
day = ['A', 'B', 'C', 'D']
sales = [40, 70, 30, 90]
plt.figure(figsize=(4,4))
plt.bar(day, sales, color='green', edgecolor='black')
plt.xlabel('day')
plt.ylabel('Sales')
plt.title('compare')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0, 10)
plt.figure(figsize=(8,8))
plt.subplot(2, 2, 1)
plt.plot(x, x, marker='o', color='red')
plt.title('Line Plot')
plt.grid(True)
plt.subplot(2, 2, 2)
plt.bar(['A','B','C'], [3,7,5], color='purple')
plt.title('Bar Chart')
plt.subplot(2, 2, 3)
plt.scatter(x, np.random.randint(1, 20, size=10), color='blue')
plt.title('Scatter Plot')
plt.subplot(2, 2, 4)
plt.pie([10,20,30], labels=['X','Y','Z'])
plt.title('Pie Chart')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(5)
data = np.random.randn(200)
plt.figure(figsize=(4,4))
plt.hist(data, bins=20, edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Normal Distribution')
plt.grid(True)
plt.show()
```
